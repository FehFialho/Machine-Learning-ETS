### Horas de Sono

```python
male = df[df["Gender"] == 'male']
female = df[df["Gender"] == 'female']


sleep_male = male["Sleep_time_hours"].value_counts().sort_index()
sleep_male.plot(kind="line", xlabel="Horas de sono", ylabel="Número de gatos", title="Gatos por horas de sono", label="Machos")

sleep_female = female["Sleep_time_hours"].value_counts().sort_index()
sleep_female.plot(kind="line", xlabel="Horas de sono", ylabel="Número de gatos", title="Gatos por horas de sono", label="Femêas")

plt.legend() 
```

### Quantidade de Cada Raça

```python
breeds = df["Breed"].value_counts().sort_index()
breeds.plot(kind="bar", xlabel="Raça", ylabel="Quantidade", title="Quantidade de gatos por raça")
```

## O que é

O Label Encoding converte valores categóricos em números inteiros.

Ele substitui cada categoria por um número (0, 1, 2, ...), mantendo tudo na mesma coluna.

## Quando Usar

O Label Encoding é recomendado para categorias que possuem ordem lógica, como:

pequeno → médio → grande

básico → intermediário → avançado

Isso porque alguns algoritmos interpretam os números como se 2 fosse “maior” que 0, o que faz sentido apenas quando há uma ordem natural nos dados.

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

# Exemplo de DataFrame
data = {
    "Tamanho": ["Pequeno", "Médio", "Grande", "Médio"],  # ordenável
    "Nivel_Escolaridade": ["Fundamental", "Médio", "Superior", "Médio"],  # ordenável
    "Idade": [25, 32, 40, 29]
}

df = pd.DataFrame(data)

# Criando o LabelEncoder
encoder = LabelEncoder()

# Aplicando o LabelEncoder em colunas ordenáveis
df["Tamanho_encoded"] = encoder.fit_transform(df["Tamanho"])
df["Nivel_encoded"] = encoder.fit_transform(df["Nivel_Escolaridade"])

print(df)
```

## O que é

O One-Hot Encoding é um método de transformação usado para converter dados categóricos não numéricos em valores numéricos binários que os algoritmos de Machine Learning conseguem entender.

Transforma cada categoria em uma nova coluna, com valores 0 ou 1 indicando se aquela linha pertence ou não à categoria.

Cada categoria vira uma coluna, e isso evita criar relações numéricas incorretas entre elas.

## Quando Usar

Para quando as categorias não possuem uma ordem lógica.

Modelos que gostam de One-hot:  
•	Árvores (Decision Tree, Random Forest)  
•	Redes neurais   
•	Regressão logística     
•	SVM (kernel linear) 



```python
# ==== Importação ====
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# ==== Base de dados exemplo ====
dados = pd.DataFrame({
    "Cor": ["Azul", "Verde", "Vermelho", "Azul", "Verde"]
})

print("=== Dados Originais ===")
print(dados)

# ==== Criando o OneHotEncoder ====
ohe = OneHotEncoder(
    sparse_output=False,   # retorna um array normal, não matriz esparsa
)

# ==== Aplicando o One Hot Encoding ====
ohe_array = ohe.fit_transform(dados[["Cor"]])

print("\n=== Resultado do One-Hot (Array Numérico) ===")
print(ohe_array)

# ==== Convertendo para DataFrame com nomes das colunas ====
colunas = ohe.get_feature_names_out(["Cor"])
ohe_df = pd.DataFrame(ohe_array, columns=colunas)

print("\n=== Resultado em DataFrame ===")
print(ohe_df)

# ==== Concatenando com o DataFrame original (se quiser manter tudo) ====
dados_final = pd.concat([dados, ohe_df], axis=1)

print("\n=== DataFrame Original + One-Hot ===")
print(dados_final)
```

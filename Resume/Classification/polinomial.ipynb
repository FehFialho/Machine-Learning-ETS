## Fórmula

A Regressão Linear Polinomial utiliza a mesma ideia da equação da reta, porém expande as variáveis originais criando termos elevados a potências (x², x³, etc.).
Isso permite modelar relações curvas, mas o modelo continua sendo linear em relação aos coeficientes.

Exemplo para grau 2:
y = b + m1·x + m2·x²

## Características

Modela relações curvas entre X e Y.

- É considerada um modelo linear porque continua linear nos coeficientes.    
- Permite capturar padrões não lineares sem trocar de algoritmo.    
- O transformador polinomial cria novas features automaticamente.   

Com degree = 2 → cria x e x²    
Com degree = 3 → cria x, x² e x³

## Hiperparâmetro

degree

Define o grau do polinômio.     
- Ex.: degree = 2 cria colunas elevadas ao quadrado.  
- Graus mais altos aumentam a complexidade e o risco de overfitting.

## Funcionamento

1. Carrega os dados (X e Y).    
2. Aplica PolynomialFeatures para gerar as colunas polinomiais. 
3. Separa os dados em treino e teste.   
4. Treina um modelo LinearRegression usando as features expandidas. 
5. Realiza previsões com o modelo treinado. 

## Código

```python
# ===============================
# 1. Criar base de dados local
# ===============================
from sklearn.datasets import make_regression  # import usado apenas aqui

X, y = make_regression(
    n_samples=1000,     # quantidade de amostras (linhas)
    n_features=1,       # apenas 1 feature (polinomial funciona melhor visualmente)
    noise=20,           # adiciona ruído aleatório para simular imperfeição dos dados
    random_state=42     # garante que sempre gere os mesmos números
)


# ===============================
# 2. Criar colunas polinomiais
# ===============================
from sklearn.preprocessing import PolynomialFeatures  # import usado apenas aqui

poly = PolynomialFeatures(
    degree=2,           # grau do polinômio (x, x²)
    include_bias=False  # mantém apenas as features criadas (evita coluna de 1)
)

X_poly = poly.fit_transform(X)  # cria as novas colunas polinomiais


# ===============================
# 3. Separar treino e teste
# ===============================
from sklearn.model_selection import train_test_split  # import usado apenas aqui

X_train, X_test, y_train, y_test = train_test_split(
    X_poly, y,
    test_size=0.2,      # 20% dos dados vão para teste
    random_state=42     # garante sempre a mesma separação
)


# ===============================
# 4. Criar e treinar o modelo
# ===============================
from sklearn.linear_model import LinearRegression  # import usado apenas aqui

modelo = LinearRegression()          # cria o modelo de Regressão Linear Polinomial
modelo.fit(X_train, y_train)         # treina usando as colunas polinomiais


# ===============================
# 5. Fazer previsões
# ===============================
y_pred = modelo.predict(X_test)      # previsões com o modelo treinado


# ===============================
# 6. Calcular MAE
# ===============================
from sklearn.metrics import mean_absolute_error  # import acima do uso

mae = mean_absolute_error(y_test, y_pred)        # mede o erro médio entre real e previsto


# ===============================
# 7. Score (R²) — treino e teste
# ===============================
scoreTrain = modelo.score(X_train, y_train)
scoreTest  = modelo.score(X_test, y_test)


# ===============================
# 8. Exibir resultados
# ===============================
print("Coeficientes:", modelo.coef_)             # coeficientes para x e x²
print("Intercepto:", modelo.intercept_)

print("\nPrimeiras previsões:", y_pred[:5])

print("\nMAE (Erro Absoluto Médio):", mae)
print("Score Treino (R²):", scoreTrain)
print("Score Teste  (R²):", scoreTest)
```

## Fórmula

O SVM procura encontrar o hiperplano que melhor separa as classes, maximizando a margem, que é a distância entre o hiperplano e os pontos mais próximos (os support vectors).
A decisão final é feita avaliando de que lado do hiperplano o novo ponto se encontra.

## Características

- É um algoritmo poderoso para problemas de classificação e regressão, especialmente quando as classes não são linearmente separáveis.

- Utiliza margens máximas, garantindo uma separação mais robusta entre classes.

- Pode trabalhar com dados não lineares através dos kernels, que transformam o espaço dos dados.

- Funciona bem em bases pequenas e médias.

- Pode ser sensível à escala, portanto normalizar/ padronizar os dados é recomendado.

- Quando usado com kernels complexos, pode ter custo computacional alto em bases grandes.

## Hiperparâmetros

kernel — tipo de função usada para separar os dados (linear, poly, rbf, sigmoid).

C — controla o quão rígida é a margem (trade-off entre margem larga e poucos erros).

gamma — afeta a curvatura da fronteira de decisão (para kernels rbf, poly e sigmoid).

## Funcionamento

• O SVM encontra o hiperplano que melhor separa as classes. 
• Se os dados não forem separáveis por uma linha/hiperplano, usa kernels para criar separação não linear.   
• Processo: criar → treinar → prever.   

```python
# ===== 1. Gerar base =====
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=800,
    n_features=4,
    n_informative=3,
    n_redundant=1,
    n_classes=2,
    random_state=42
)

# ===== 2. Treino / Teste =====
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# ===== 3. Modelo SVM =====
from sklearn.svm import SVC

model = SVC(kernel="rbf")  # kernel padrão

model.fit(X_train, y_train)

# ===== 4. Predição e Métrica =====
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test)
acc = accuracy_score(y_test, y_pred)

print("Accuracy:", acc)

# ===== 5. ConfusionMatrix (Yellowbrick) =====
from yellowbrick.classifier import ConfusionMatrix

visualizer = ConfusionMatrix(model)

visualizer.fit(X_train, y_train)
visualizer.score(X_test, y_test)
visualizer.show()
```

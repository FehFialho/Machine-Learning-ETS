### Fórmula
Assim como a regressão linear comum, a regressão linear múltipla utiliza da equação da reta, mas considerando um valor de inclinação para cada variável exploratória.

### Características
- Busca encontrar a linha reta que melhor representa o comportamento de um conjunto de dados.   
- Diferentemente da regressão linear simples, incorpora várias variáveis independentes no mesmo modelo.    
- Permite uma análise mais completa e precisa das relações existentes.     
- Cada variável contribui para o resultado com um peso diferente (coeficientes).

### Hiperparâmetro
Nenhum.

### Funcionamento
Usa um hiperplano para prever valores numéricos.    
• Encontra os melhores coeficientes que representam a relação entre X (múltiplas entradas) e Y (saída).   
• Processo: carregar dados → separar treino/teste → treinar → prever valores.  

## Código

```python
# ===============================
# 1. Criar base de dados local
# ===============================
from sklearn.datasets import make_regression  # import usado apenas aqui

X, y = make_regression(
    n_samples=1000,     # quantidade de amostras (linhas)
    n_features=3,       # número de variáveis independentes (múltiplas features)
    noise=20,           # adiciona ruído aleatório para simular imperfeição dos dados
    random_state=42     # garante que sempre gere os mesmos números
)


# ===============================
# 2. Separar treino e teste
# ===============================
from sklearn.model_selection import train_test_split  # import usado apenas aqui

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% dos dados vão para teste
    random_state=42     # garante sempre a mesma separação
)


# ===============================
# 3. Criar e treinar o modelo
# ===============================
from sklearn.linear_model import LinearRegression  # import usado apenas aqui

modelo = LinearRegression()   # cria o modelo de Regressão Linear Múltipla
modelo.fit(X_train, y_train)  # treina usando apenas os dados de treino


# ===============================
# 4. Fazer previsões
# ===============================
# previsões são feitas usando o modelo já treinado
y_pred = modelo.predict(X_test)


# ===============================
# 5. Calcular MAE
# ===============================
from sklearn.metrics import mean_absolute_error  # import acima do uso

mae = mean_absolute_error(y_test, y_pred)  # mede o erro médio entre real e previsto


# ===============================
# 6. Score (R²) — treino e teste
# ===============================
# (score é do próprio modelo — não precisa importar nada)

scoreTrain = modelo.score(X_train, y_train)
scoreTest  = modelo.score(X_test, y_test)     


# ===============================
# 7. Exibir resultados
# ===============================
print("Intercepto:", modelo.intercept_)
print("Coeficientes:", modelo.coef_)  # um coeficiente para cada feature

print("\nPrimeiras previsões:", y_pred[:5])

print("\nMAE (Erro Absoluto Médio):", mae)
print("Score Treino (R²):", scoreTrain)
print("Score Teste (R²):", scoreTest)
```
